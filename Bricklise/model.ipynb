Let's use logistic regression to estimate if the patient is sick or not.
y= -1 (sick), y= 1 (healthy)

```python
# Imports
import numpy as np
from implementations1 import *
from sklearn.model_selection import train_test_split
```

Load data

```python
from helpers import load_csv_data, create_csv_submission

# load data.
x_train, x_test, y_train, train_ids, test_ids = load_csv_data(data_path = "Bricklise\\data")

# split the training data into training and validation set
x_train, x_val, y_train, y_val = train_test_split(
    x_train,
    y_train,
    test_size=0.2,
    random_state=42
)
# try later with several training and validation sets
```

Newtons method logistic regression

```python
# Train
initial_w = np.zeros((x_train.shape[1], 1))

loss, w = logistic_regression(y_train, x_train, initial_w, max_iters=1000, gamma=0.1)
```

```python
# Predict probabilities
y_val_prob = sigmoid(x_val @ w)

# Convert probabilities to predictions
y_val_pred = np.where(y_val_prob >= 0.5, 1, -1)
```

```python
accuracy = np.mean(y_val_pred.flatten() == y_val)

print("Validation accuracy:", accuracy)
```
