#### Changer les proportions de data de malades/ non malades

```python
def balance_dataset(x_train, y_train, positive_percentage):
    # Keep only the label column, not the whole DataFrame
    y_values = y_train.iloc[:, 1].to_numpy()

    indices_negative = np.flatnonzero(y_values == -1)
    indices_positive = np.flatnonzero(y_values == 1)

    if not (0 < positive_percentage < 1):
        raise ValueError("positive_percentage must be between 0 and 1")

    scale = min(
        len(indices_positive) / positive_percentage,
        len(indices_negative) / (1 - positive_percentage)
    )
    n_positive = int(scale * positive_percentage)
    n_negative = int(scale * (1 - positive_percentage))

    rng = np.random.default_rng(42)
    selected_positive = rng.choice(indices_positive, size=n_positive, replace=False)
    selected_negative = rng.choice(indices_negative, size=n_negative, replace=False)

    selected_indices = np.concatenate([selected_positive, selected_negative])
    rng.shuffle(selected_indices)

    x_balanced = x_train.iloc[selected_indices].reset_index(drop=True)
    y_balanced = y_train.iloc[selected_indices].reset_index(drop=True)

    return x_balanced, y_balanced
```

Enregistrer un nouveau csv avec un pourcetnage de 1, -1 different

```python
x_train_balanced_30, y_train_balanced_30 = balance_dataset(x_train, y_train, positive_percentage=0.3)
```

```python
#############################
# VERIFICATION
#############################

# Prendre uniquement la deuxième colonne de y_train
y = y_train_balanced_30.iloc[:, 1]

# Compter les -1 et les 1
n_negative = np.sum(y == -1)
n_positive = np.sum(y == 1)

# Total
n_total = n_negative + n_positive

# Proportions
proportion_negative = n_negative / n_total
proportion_positive = n_positive / n_total

# Affichage
print("Nombre de -1 :", n_negative)
print("Nombre de  1 :", n_positive)

print("Proportion de -1 :", f"{proportion_negative * 100:.2f}%")
print("Proportion de  1 :", f"{proportion_positive * 100:.2f}%")
```
